1. Basic setup — one cell

```python
from byteflow import Agent
from byteflow.providers.ollama_provider import OllamaProvider

agent = Agent(provider=OllamaProvider(), memory_path="notebook_memory.json")
```

2. Plain chat

```python
response = agent.chat("Explain how binary search works")
print(response)
```

3. Tool calling / math

```python
from byteflow.builtin_tools import register_builtin_tools

register_builtin_tools(agent)  # adds add/subtract/multiply/divide
print(agent.run("add 10 and 20"))
```

4. Coding mode — generate and run code

```python
result = agent.code("write a function that checks if a number is prime", execute=True)

print(result["code"])
print(result["result"].format())
```

5. Desktop tools (optional)


```python
from byteflow.desktop_tools import register_desktop_tools

register_desktop_tools(agent)
print(agent.use_tool("launch", "youtube"))
print(agent.use_tool("list_files", r"C:\Users\aman\Videos", "*.pdf"))
```

6. Learned profile / facts

```python
print(agent.profile.format())   # see what it's learned about you so far
```

7. Drafting a social post

```python
result = agent.draft_social_post("launching my project", platform="linkedin", tone="excited")
print(result["draft"])
```
