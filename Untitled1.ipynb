```python
from bs4 import BeautifulSoup
import pandas as pd
import re

# Read HTML file
with open("FLIPKART_TV.html", "r", encoding="utf-8") as file:
    soup = BeautifulSoup(file, "html.parser")
print(soup.get_text()[:3000])
# Lists to store data
names = []
sizes = []
stars = []
reviews = []
actual_prices = []
final_prices = []
launch_years = []
discounts = []

# Find all product blocks
products = soup.find_all("div", class_="_1AtVbE")

for product in products:

    # Product Name
    name_tag = product.find("div", class_="_4rR01T")
    if not name_tag:
        name_tag = product.find("a", class_="s1Q9rs")

    if name_tag:
        name = name_tag.get_text(strip=True)
    else:
        continue

    # Extract size
    size_match = re.search(r'(\d+\s*inch)', name, re.I)
    size = size_match.group(1) if size_match else None

    # Extract launch year
    year_match = re.search(r'(202[0-9])', name)
    launch_year = year_match.group(1) if year_match else None

    # Rating
    star_tag = product.find("div", class_="_3LWZlK")
    stars_val = star_tag.text if star_tag else None

    # Reviews
    review_tag = product.find("span", class_="_2_R_DZ")
    if review_tag:
        review_match = re.search(r'([\d,]+)', review_tag.text)
        review_val = review_match.group(1) if review_match else None
    else:
        review_val = None

    # Final Price
    final_price_tag = product.find("div", class_="_30jeq3")
    final_price = final_price_tag.text.replace("₹", "") if final_price_tag else None

    # Actual Price
    actual_price_tag = product.find("div", class_="_3I9_wc")
    actual_price = actual_price_tag.text.replace("₹", "") if actual_price_tag else None

    # Discount
    discount_tag = product.find("div", class_="_3Ay6Sb")
    discount = discount_tag.text if discount_tag else None

    names.append(name)
    sizes.append(size)
    stars.append(stars_val)
    reviews.append(review_val)
    actual_prices.append(actual_price)
    final_prices.append(final_price)
    launch_years.append(launch_year)
    discounts.append(discount)

# Create DataFrame
df = pd.DataFrame({
    "Name": names,
    "Size": sizes,
    "Stars": stars,
    "Reviews": reviews,
    "Actual Price": actual_prices,
    "Final Price": final_prices,
    "Launch Year": launch_years,
    "Discount": discounts
})

print(df)

# Save to CSV
df.to_csv("Flipkart_TV_Data.csv", index=False)

print("\nData extracted successfully!")
```

```python
from bs4 import BeautifulSoup
import pandas as pd
import re

# Read HTML file
with open("FLIPKART_TV.html", "r", encoding="utf-8") as f:
    soup = BeautifulSoup(f, "html.parser")

data = []

# Find all text-containing elements
for tag in soup.find_all(["div", "a"]):

    text = tag.get_text(" ", strip=True)

    # Check if it looks like a TV product
    if ("TV" in text or "Smart TV" in text) and "₹" in text and len(text) < 500:
        # Product Name
        name = text[:100]

        # Size
        size_match = re.search(r'(\d+)\s*inch', text, re.I)
        size = size_match.group(1) + " inch" if size_match else ""

        # Rating
        rating_match = re.search(r'(\d\.\d)', text)
        rating = rating_match.group(1) if rating_match else ""

        # Prices
        prices = re.findall(r'₹[\d,]+', text)

        final_price = prices[0] if len(prices) >= 1 else ""
        actual_price = prices[1] if len(prices) >= 2 else ""

        # Discount
        discount_match = re.search(r'(\d+% ?off)', text, re.I)
        discount = discount_match.group(1) if discount_match else ""

        # Launch Year
        year_match = re.search(r'20\d{2}', text)
        year = year_match.group() if year_match else ""

        data.append([
            name,
            size,
            rating,
            actual_price,
            final_price,
            year,
            discount
        ])

# Create DataFrame
df = pd.DataFrame(
    data,
    columns=[
        "Name",
        "Size",
        "Stars",
        "Actual Price",
        "Final Price",
        "Launch Year",
        "Discount"
    ]
)

# Remove duplicates
df = df.drop_duplicates()

print(df)

# Save CSV
df.to_csv("Flipkart_TV_Data.csv", index=False)

print("\nData extracted successfully!")
```

```python
from bs4 import BeautifulSoup
import pandas as pd
import re

# Open HTML File
with open("FLIPKART_TV.html", "r", encoding="utf-8") as file:
    soup = BeautifulSoup(file, "html.parser")

# Lists
names = []
sizes = []
stars = []
reviews = []
actual_prices = []
final_prices = []
launch_years = []
discounts = []

# Extract Product Information
for text in soup.stripped_strings:

    if "TV" in text and "inch" in text and len(text) > 40:

        # Name
        name = text

        # Size
        size_match = re.search(r'(\d+)\s*inch', text)
        size = size_match.group(1) + " inch" if size_match else ""

        # Launch Year
        year_match = re.search(r'20\d{2}', text)
        launch_year = year_match.group() if year_match else ""

        # Rating
        rating_match = re.search(r'(\d\.\d)', text)
        star = rating_match.group(1) if rating_match else ""

        # Reviews
        review_match = re.search(r'([\d,]+)\s*Ratings', text)
        review = review_match.group(1) if review_match else ""

        # Prices
        price_match = re.findall(r'₹[\d,]+', text)

        final_price = price_match[0] if len(price_match) > 0 else ""
        actual_price = price_match[1] if len(price_match) > 1 else ""

        # Discount
        discount_match = re.search(r'(\d+% ?off)', text, re.IGNORECASE)
        discount = discount_match.group(1) if discount_match else ""

        # Store Data
        names.append(name)
        sizes.append(size)
        stars.append(star)
        reviews.append(review)
        actual_prices.append(actual_price)
        final_prices.append(final_price)
        launch_years.append(launch_year)
        discounts.append(discount)

# Create DataFrame
df = pd.DataFrame({
    "Name": names,
    "Size": sizes,
    "Stars": stars,
    "Reviews": reviews,
    "Actual Price": actual_prices,
    "Final Price": final_prices,
    "Launch Year": launch_years,
    "Discount": discounts
})

# Remove Duplicate Rows
df = df.drop_duplicates()

# Display Data
print(df)

# Save CSV File
df.to_csv("Flipkart_TV_Data.csv", index=False)

print("\nData extracted successfully!")
```

```python
import os
import re
import pandas as pd
from bs4 import BeautifulSoup

# Define the file path
html_file_path = "FLIPKART_TV.html"

# Verify file existence
if not os.path.exists(html_file_path):
    raise FileNotFoundError(f"The required file '{html_file_path}' was not found. Please ensure it is in the working directory.")

# Read and parse the HTML file
with open(html_file_path, "r", encoding="utf-8") as f:
    soup = BeautifulSoup(f.read(), "html.parser")

# Target lists to capture each column's data
names = []
sizes = []
stars = []
reviews = []
actual_prices = []
final_prices = []
launch_years = []
discounts = []

# Find all container blocks containing product details on Flipkart
# Usually, product rows in list views are encapsulated in a div with a specific class or structure.
# We look for containers holding titles/prices.
product_cards = soup.find_all("div", class_="tUxRFH")

for card in product_cards:
    # 1. Extract Full Title / Name
    name_tag = card.find("div", class_="KzDlHZ")
    full_name = name_tag.text.strip() if name_tag else "NaN"
    
    # Clean name and extract size where applicable
    # Example raw: "LG UR7500 126 cm (50 inch) Ultra HD (4K) LED..."
    # Target Expected Output: "LG UR7500 126 cm" and "50 inch"
    size_match = re.search(r'\((\d+\s*inch)\)', full_name, re.IGNORECASE)
    if size_match:
        size_str = size_match.group(1)
        # Strip out the bracket details to match the concise output layout
        cleaned_name = re.sub(r'\s*\(.*\).*$', '', full_name).strip()
    else:
        size_str = "NaN"
        cleaned_name = full_name
        
    names.append(cleaned_name)
    sizes.append(size_str)
    
    # 2. Extract Star Rating
    star_tag = card.find("div", class_="XQDdHH")
    stars.append(float(star_tag.text.strip()) if star_tag else "NaN")
    
    # 3. Extract Reviews Count from text structure
    review_container = card.find("span", class_="Wphh3N")
    review_count = "NaN"
    if review_container:
        # Expected format: "21,843 Ratings & 1,631 Reviews"
        text = review_container.text
        review_match = re.search(r'&\s*([\d,]+)\s*Reviews', text)
        if review_match:
            review_count = review_match.group(1)
    reviews.append(review_count)
            
    # 4. Extract Pricing
    final_price_tag = card.find("div", class_="Nx9bqj")
    final_price = final_price_tag.text.replace("₹", "").strip() if final_price_tag else "NaN"
    final_prices.append(final_price)
    
    actual_price_tag = card.find("div", class_="yRaY8j")
    actual_price = actual_price_tag.text.replace("₹", "").strip() if actual_price_tag else "NaN"
    actual_prices.append(actual_price)
    
    # 5. Extract Discount Percentage
    discount_tag = card.find("div", class_="UkUFwK")
    discount_val = "NaN"
    if discount_tag:
        discount_match = re.search(r'(\d+%)', discount_tag.text)
        if discount_match:
            discount_val = discount_match.group(1)
    discounts.append(discount_val)
            
    # 6. Extract Launch Year from specifications list
    launch_year = "NaN"
    specs_list = card.find("ul", class_="G4BRas")
    if specs_list:
        for li in specs_list.find_all("li"):
            if "Launch Year" in li.text:
                launch_year = li.text.replace("Launch Year:", "").strip()
                break
    launch_years.append(launch_year)

# Structure extracted features into a Pandas DataFrame
df = pd.DataFrame({
    "Name": names,
    "Size": sizes,
    "Stars": stars,
    "Reviews": reviews,
    "Actual Price": actual_prices,
    "Final Price": final_prices,
    "Launch Year": launch_years,
    "Discount": discounts
})

# Display the final DataFrame
print(df.to_string())
```
